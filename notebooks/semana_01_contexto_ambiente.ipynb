## 2. Diagnóstico de Aprendajes Previos

1. **Dato vs. Información:** Un *dato* es un registro o cifra en bruto sin procesar (ej. 68%). La *información* es el dato contextualizado y organizado que genera un significado (ej. asistencia del 68% en la ficha de formación).
2. **Ejemplos en un ambiente de formación:** Porcentaje de asistencia a clases y calificaciones obtenidas en las evidencias de aprendizaje.
3. **Archivo CSV:** Formato de texto plano con valores separados por comas, utilizado para almacenar datos tabulares de forma liviana y portable.
4. **Fila vs. Columna:** Una *fila* (registro) representa una entidad o sujeto individual (ej. un aprendiz); una *columna* (variable/atributo) representa una característica medida para todos los registros (ej. nota o asistencia).
5. **Tipo de dato para nota:** Número decimal (`float`), ya que permite registrar decimales (ej. 3.8, 4.5).
6. **Riesgo de datos personales:** Vulnerar la privacidad de las personas, incumplir la ley de Habeas Data (protección de datos) y generar sesgos de discriminación.
7. **Pregunta sobre tabla:** *"¿Cuál es la nota promedio general por ficha de formación?"*
8. **Herramientas conocidas:** Experiencia con Python, Excel, SQL y editores de código como Visual Studio Code.
9. **Dato incompleto vs. Duplicado:** Un *dato incompleto* indica la ausencia de un valor (nulo / `NaN`); un *dato duplicado* es un registro idéntico que se repite más de una vez en el conjunto de datos.
10. **Expectativa:** Aprender a procesar, limpiar, analizar y visualizar datos con Python para apoyar la toma de decisiones fundamentadas en evidencia.
