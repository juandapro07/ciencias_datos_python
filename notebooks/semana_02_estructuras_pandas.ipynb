## 2. Estructuras Básicas: Listas y Diccionarios

A continuación representamos los datos iniciales de los aprendices mediante una lista de diccionarios, incluyendo la resolución del reto del aprendiz.

```python
# Lista inicial de diccionarios con aprendices
datos_aprendices = [
    {"documento": "1001", "nombre": "Ana", "programa": "ADSO", "nota_python": 4.2, "asistencia": 92},
    {"documento": "1002", "nombre": "Carlos", "programa": "ADSO", "nota_python": 3.8, "asistencia": 85},
    {"documento": "1003", "nombre": "María", "programa": "ADSO", "nota_python": 4.7, "asistencia": 96},
]

# RETO DEL APRENDIZ: Agregar dos nuevos registros
datos_aprendices.append({"documento": "1004", "nombre": "Luis", "programa": "ADSO", "nota_python": 2.9, "asistencia": 70})
datos_aprendices.append({"documento": "1005", "nombre": "Sofía", "programa": "ADSO", "nota_python": 5.0, "asistencia": 98})

print(f"Total de registros cargados: {len(datos_aprendices)}")
datos_aprendices
```

## 3. NumPy Básico para Arreglos Numéricos

Utilizamos NumPy para realizar operaciones vectorizadas, estadísticas descriptivas básicas y aplicar filtros condicionales sobre las notas.


```python
import numpy as np

# Creación de arreglo numérico con notas
notas = np.array([4.2, 3.8, 4.7, 2.9, 5.0])

print("--- Estadísticas Básicas con NumPy ---")
print("Notas:", notas)
print("Promedio:", np.mean(notas))
print("Mediana:", np.median(notas))
print("Máxima:", np.max(notas))
print("Mínima:", np.min(notas))

# Operación vectorizada: Bonificación controlada de 0.2 sin superar 5.0
notas_bonificadas = np.clip(notas + 0.2, 0, 5)
print("\nNotas Bonificadas:", notas_bonificadas)

# Filtro condicional: Aprobadas (>= 3.5)
notas_aprobadas = notas[notas >= 3.5]
porcentaje_aprobacion = (len(notas_aprobadas) / len(notas)) * 100
print(f"\nNotas aprobadas: {notas_aprobadas}")
print(f"Porcentaje de aprobación: {porcentaje_aprobacion:.1f}%")
```

## 4. Introducción a Pandas: Series y DataFrames

Construimos un DataFrame a partir de la lista de diccionarios y agregamos columnas calculadas con condiciones vectorizadas.

```python
import pandas as pd

# Creación de un DataFrame
df_aprendices = pd.DataFrame(datos_aprendices)

# Agregar columna calculada de 'estado'
df_aprendices["estado"] = np.where(
    df_aprendices["nota_python"] >= 3.5,
    "Aprobado",
    "En riesgo"
)

# RETO DEL APRENDIZ: Columna 'alerta_asistencia' (< 80)
df_aprendices["alerta_asistencia"] = np.where(
    df_aprendices["asistencia"] < 80,
    "Revisar asistencia",
    "Sin alerta"
)

df_aprendices
```

## 5. Importación de Datos CSV y Excel con Rutas Relativas

Creamos un dataset base de prueba, lo exportamos a la carpeta `data/raw` y posteriormente lo importamos usando `pandas`.    

```python
from pathlib import Path

# Definición de rutas relativas
BASE_DIR = Path("..")
DATA_RAW = BASE_DIR / "data" / "raw"
DATA_PROCESSED = BASE_DIR / "data" / "processed"
OUTPUTS = BASE_DIR / "outputs"

DATA_RAW.mkdir(parents=True, exist_ok=True)
DATA_PROCESSED.mkdir(parents=True, exist_ok=True)
OUTPUTS.mkdir(parents=True, exist_ok=True)

# Dataset base ampliado
datos_base = {
    "documento": ["1001", "1002", "1003", "1004", "1005"],
    "nombre": ["Ana", "Carlos", "María", "Luis", "Sofía"],
    "programa": ["ADSO", "ADSO", "ADSO", "ADSO", "ADSO"],
    "ficha": ["287001", "287001", "287001", "287001", "287001"],
    "nota_python": [4.2, 3.8, 4.7, 2.9, 5.0],
    "nota_datos": [4.0, 3.6, 4.8, 3.1, 4.9],
    "asistencia": [92, 85, 96, 70, 98],
    "ciudad": ["Medellín", "Bello", "Itagüí", "Medellín", "Envigado"]
}

df_base = pd.DataFrame(datos_base)

# Exportación inicial a data/raw
ruta_csv = DATA_RAW / "seguimiento_aprendices_semana2.csv"
ruta_excel = DATA_RAW / "seguimiento_aprendices_semana2.xlsx"

df_base.to_csv(ruta_csv, index=False, encoding="utf-8")
df_base.to_excel(ruta_excel, index=False)

# Carga de comprobación desde data/raw
df_csv = pd.read_csv(ruta_csv)
df_excel = pd.read_excel(ruta_excel)

print("✅ Archivos leídos correctamente:")
print("CSV Shape:", df_csv.shape)
print("Excel Shape:", df_excel.shape)
```

## 6. Primer Informe Exploratorio del Dataset

Aplicamos los comandos de inspección inicial para examinar la estructura, dimensiones y propiedades de las columnas.


```python
df_seguimiento = pd.read_csv(DATA_RAW / "seguimiento_aprendices_semana2.csv")

print("--- 1. Primeros 3 Registros (head) ---")
display(df_seguimiento.head(3))

print("\n--- 2. Últimos 2 Registros (tail) ---")
display(df_seguimiento.tail(2))

print("\n--- 3. Dimensiones (shape) ---")
print("Filas y Columnas:", df_seguimiento.shape)

print("\n--- 4. Nombres de Columnas (columns) ---")
print(df_seguimiento.columns.tolist())

print("\n--- 5. Tipos de Datos (dtypes) ---")
print(df_seguimiento.dtypes)

print("\n--- 6. Información General (info) ---")
df_seguimiento.info()

print("\n--- 7. Estadísticas Descriptivas (describe) ---")
display(df_seguimiento.describe())
```

### Hallazgos Iniciales
- **Dimensiones:** El dataset tiene **5 filas** y **8 columnas**.
- **Columnas Numéricas:** `nota_python` (float), `nota_datos` (float), `asistencia` (int).
- **Columnas de Texto:** `documento`, `nombre`, `programa`, `ficha`, `ciudad`.
- **Columna a revisar en Semana 3:** La columna `asistencia` para verificar alertas de riesgo por debajo del 80%.
- **Pregunta analítica:** ¿Cuál es la relación entre el porcentaje de asistencia y el promedio general de calificaciones por aprendiz?

## 7. Exportación y Cierre Técnico

Calculamos el promedio general, asignamos el estado académico y exportamos el dataset procesado a la carpeta `data/processed/`.

```python
# Cálculo de columnas procesadas
df_seguimiento["promedio_general"] = (
    df_seguimiento["nota_python"] + df_seguimiento["nota_datos"]
) / 2

df_seguimiento["estado_academico"] = np.where(
    df_seguimiento["promedio_general"] >= 3.5,
    "Aprobado",
    "En riesgo"
)

# Selección y ordenamiento de columnas finales
df_final = df_seguimiento[[
    "documento", "nombre", "programa", "ficha",
    "nota_python", "nota_datos", "promedio_general",
    "asistencia", "estado_academico", "ciudad"
]]

# Exportar a data/processed
ruta_csv_proc = DATA_PROCESSED / "seguimiento_semana2_procesado.csv"
ruta_excel_proc = DATA_PROCESSED / "seguimiento_semana2_procesado.xlsx"

df_final.to_csv(ruta_csv_proc, index=False, encoding="utf-8")
df_final.to_excel(ruta_excel_proc, index=False)

print("✅ Dataset procesado exportado exitosamente en:")
print("•", ruta_csv_proc)
print("•", ruta_excel_proc)

display(df_final.head())
```
